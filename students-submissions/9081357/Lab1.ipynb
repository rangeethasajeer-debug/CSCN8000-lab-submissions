## 5.Create two 2x2 arrays and demonstrate how you can stack the elements vertically, horizontally, and split the arrays into smaller arrays.

```python


#create two 2x2 arrays
array1 = np.array([[10, 20], [30, 40]])
array2 = np.array([[50, 60], [70, 80]])

print("Array 1:\n", array1)
print("\nArray 2:\n", array2)

#Vertical stacking
v_stacked_array = np.vstack((array1, array2))
print("\nVertically Stacked Array:\n", v_stacked_array)

#Horizontal stacking
h_stacked_array = np.hstack((array1, array2))
print("\nHorizontally Stacked Array:\n", h_stacked_array)

#Split the vertically stacked arrays in to two original 2x2 arrays
split_v_arrays = np.split(v_stacked_array, 2)
print("\nSplitting Vertically Stacked Array (array1, array2):\n")
print("Split Array 1:\n", split_v_arrays[0])
print("Split Array 2:\n", split_v_arrays[1])

#Split the horizontally stacked arrays into two original 2x2 arrays
split_h_arrays = np.hsplit(h_stacked_array, 2)
print("\nSplitting Horizontally Stacked Array (array1, array2):\n")
print("Split Array 1:\n", split_h_arrays[0])
print("Split Array 2:\n", split_h_arrays[1])





```


## 8. How does broadcasting work? Demonstrate the subtraction, multiplication by considering a 3 x 3 matrix.


### NumPy Broadcasting Explained

NumPy's broadcasting is a powerful mechanism that allows operations between arrays with different shapes. When performing arithmetic operations (like addition, subtraction, multiplication, etc.) on arrays of different sizes, NumPy attempts to 'broadcast' the smaller array across the larger array so that they have compatible shapes. This avoids the need for creating explicit copies of arrays to match their dimensions.

**Rules of Broadcasting:**
1.  **Equal dimensions:** If the arrays have different numbers of dimensions, the shape of the one with fewer dimensions is `padded` with ones on its left side.
2.  **Compatible sizes:** For each dimension, the sizes must be either equal, or one of them must be 1.
3.  **Incompatible sizes:** If these conditions are not met, a `ValueError` is raised, indicating that the arrays cannot be broadcast together.

When broadcasting is possible, the smaller array is virtually stretched or copied to match the larger array's dimensions for the operation, but no actual memory is duplicated.

```python


# Create a 3x3 matrix
matrix_3x3 = np.array([[10, 20, 30],
                       [40, 50, 60],
                       [70, 80, 90]])
print("Original 3x3 Matrix:\n", matrix_3x3)

# --- Demonstration of Subtraction with Broadcasting ---

# Scenario 1: Subtract a scalar
scalar_value = 5
subtraction_scalar = matrix_3x3 - scalar_value
print(f"\nMatrix after subtracting scalar {scalar_value}:\n", subtraction_scalar)

# Scenario 2: Subtract a 1D array (row-wise broadcasting)
# The 1D array [1, 2, 3] will be broadcast across each row of the 3x3 matrix.
row_vector = np.array([1, 2, 3])
subtraction_row = matrix_3x3 - row_vector
print(f"\nMatrix after subtracting row vector {row_vector} (broadcasted row-wise):\n", subtraction_row)

# Scenario 3: Subtract a 1D column array (column-wise broadcasting)
# For column-wise broadcasting, the 1D array needs to be explicitly reshaped to a column vector (3,1).
col_vector = np.array([10, 20, 30]).reshape(3, 1)
subtraction_col = matrix_3x3 - col_vector
print(f"\nMatrix after subtracting column vector:\n", col_vector, "(broadcasted column-wise):\n", subtraction_col)

# --- Demonstration of Multiplication with Broadcasting ---

# Scenario 1: Multiply by a scalar
multiplication_scalar = matrix_3x3 * 2
print("\nMatrix after multiplying by scalar 2:\n", multiplication_scalar)

# Scenario 2: Multiply by a 1D array (row-wise broadcasting)
multiplication_row = matrix_3x3 * row_vector
print(f"\nMatrix after multiplying by row vector {row_vector} (broadcasted row-wise):\n", multiplication_row)

# Scenario 3: Multiply by a 1D column array (column-wise broadcasting)
multiplication_col = matrix_3x3 * col_vector
print(f"\nMatrix after multiplying by column vector:\n", col_vector, "(broadcasted column-wise):\n", multiplication_col)
```

###   1.Solve the following two system of linear equations using matrices (find the values of x1 and x2) and determine the number of solutions using the functions.

### 1. 2x1+3x2-4x3=6,x1-4x2=8

```python

import sympy

# --- Set 1: 2x1 + 3x2 - 4x3 = 6, x1 - 4x2 = 8 ---

print("--- Set1: 2x1 + 3x2 - 4x3 = 6, x1 - 4x2 = 8 ---")

# Define variables symbolically for clarity in solution
x1, x2, x3 = sympy.symbols('x1 x2 x3')

# Equations in matrix form Ax = b for rank analysis
# A represents the coefficients of x1, x2, x3
A1_rank = np.array([[2, 3, -4],
                    [1, -4, 0]])
# b represents the constants on the right side
b1_rank = np.array([6, 8])

# Augmented matrix for rank check
Ab1_rank = np.concatenate((A1_rank, b1_rank.reshape(-1, 1)), axis=1)

rank_A1 = np.linalg.matrix_rank(A1_rank);
rank_Ab1 = np.linalg.matrix_rank(Ab1_rank);
num_vars1 = A1_rank.shape[1] # Number of variables (x1, x2, x3)

print(f"Matrix A for System 1:\n{A1_rank}")
print(f"Augmented Matrix [A|b] for System 1:\n{Ab1_rank}")
print(f"Rank of coefficient matrix A: {rank_A1}")
print(f"Rank of augmented matrix [A|b]: {rank_Ab1}")
print(f"Number of variables: {num_vars1}")

if rank_A1 != rank_Ab1:
    print("Number of solutions: No solution (inconsistent system)")
elif rank_A1 == num_vars1:
    # This condition typically implies a unique solution if the matrix is square.
    # For an underdetermined system (2 equations, 3 variables), this path won't be taken for x1, x2, x3.
    print("Number of solutions: Unique solution (This scenario is not met as rank < number of variables)")
else: # rank_A1 == rank_Ab1 < num_vars1
    print("Number of solutions: Infinitely many solutions")
    print("Expressing solutions in terms of free variables:")

    # Using sympy to solve the system symbolically
    eq1 = sympy.Eq(2*x1 + 3*x2 - 4*x3, 6)
    eq2 = sympy.Eq(x1 - 4*x2, 8)

    solution = sympy.linsolve([eq1, eq2], (x1, x2, x3))
    print(f"Solution (x1, x2, x3): {solution}")
```

###  2. 3y1-4y2+5y3=10,-y1+2y2-4y3=8

```python
print("\n--- Set 2: 3y1 - 4y2 + 5y3 = 10, -y1 + 2y2 - 4y3 = 8 ---")

# Define variables symbolically
y1, y2, y3 = sympy.symbols('y1 y2 y3')

# Coefficient matrix for y1, y2, y3
A2_rank = np.array([
    [3, -4, 5],
    [-1, 2, -4]
])
# Constant vector
b2_rank = np.array([10, 8])

# Augmented matrix for rank check
Ab2_rank = np.concatenate((A2_rank, b2_rank.reshape(-1, 1)), axis=1)

rank_A2 = np.linalg.matrix_rank(A2_rank)
rank_Ab2 = np.linalg.matrix_rank(Ab2_rank)
num_vars2 = A2_rank.shape[1] # Number of variables (y1, y2, y3)

print(f"Matrix A for System 2:\n{A2_rank}")
print(f"Augmented Matrix [A|b] for System 2:\n{Ab2_rank}")
print(f"Rank of coefficient matrix A: {rank_A2}")
print(f"Rank of augmented matrix [A|b]: {rank_Ab2}")
print(f"Number of variables: {num_vars2}")

if rank_A2 != rank_Ab2:
    print("Number of solutions: No solution (inconsistent system)")
elif rank_A2 == num_vars2:
    print("Number of solutions: Unique solution (This case is unlikely for a 2x3 system)")
else: # rank_A2 == rank_Ab2 < num_vars2
    print("Number of solutions: Infinitely many solutions")
    print("Expressing solutions in terms of free variables:")

    # Using sympy to solve the system symbolically
    eq3 = sympy.Eq(3*y1 - 4*y2 + 5*y3, 10)
    eq4 = sympy.Eq(-y1 + 2*y2 - 4*y3, 8)

    solution2 = sympy.linsolve([eq3, eq4], (y1, y2, y3))
    print(f"Solution: {solution2}")
```
